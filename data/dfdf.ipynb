```python
import pandas as pd

# Load the dataset
df = pd.read_csv(
    "hf://datasets/AmaanP314/youtube-comment-sentiment/youtube-comments-sentiment.csv"
)

# Keep only the required columns
df = df[["CommentText", "Sentiment"]]

# Rename columns exactly as required
df = df.rename(columns={
    "CommentText": "text",
    "Sentiment": "label"
})

# Keep only the required labels
df = df[df["label"].isin(["Positive", "Negative", "Neutral"])]

# Remove empty comments
df = df.dropna(subset=["text", "label"])

# Remove duplicate comments
df = df.drop_duplicates(subset=["text"])

# # Save
df.to_csv("youtube_sentiment.csv", index=False)

print(df.shape)
print(df["label"].value_counts())
print(df.head())
```
